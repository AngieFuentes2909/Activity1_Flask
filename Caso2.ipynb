### **Ejemplo 1:** Programación Lineal (PL)

```
# Instalación del pequete de programación lineal PuLP
!pip install PuLP

# Importar paquete de programación lineal
import pulp
```

```
# Definición del problema (P) y tipo de optimización (min o max)
Q = pulp.LpProblem("Distribución de bienes a través de una red",  pulp.LpMinimize) # si el problema es de Minimización se debe usar pulp.LpMinimize

# Definir variables de decisión (x,y)
xF1_F2 = pulp.LpVariable("Cantidad de productos enviados desde fabrica 1 a fabrica 2", lowBound = 0, upBound = None)
xF1_CD= pulp.LpVariable("Cantidad de productos enviados desde fabrica 1 al centro de distribucion", lowBound = 0, upBound = None)
xF1_A1 = pulp.LpVariable("Cantidad de productos enviados desde fabrica 1 a almacen 1", lowBound = 0, upBound = None)
xF2_CD= pulp.LpVariable("Cantidad de productos enviados desde fabrica 2 al centro de distribucion", lowBound = 0, upBound = None)
xCD_A2= pulp.LpVariable("Cantidad de productos enviados desde Centro de distribucion a almacen 2", lowBound = 0, upBound = None)
xA1_A2= pulp.LpVariable("Cantidad de productos enviados desde almacen 1 a almacen 2", lowBound = 0, upBound = None)
xA2_A1= pulp.LpVariable("Cantidad de productos enviados desde almacen 2 a almacen 1", lowBound = 0, upBound = None)


# Definir función objetivo
def F(x,y):
  return  3600*x + 5400*y   # Ganancia por venta de ambos modelos

# Definir los recursos utilizados en función del plan de producción (x,y)
def R2(x,y):
  return  4*x + 2*y     # Cantidad de puertas

def R1(x,y):
  return  6*x + 10.5*y  # Cantidad de horas laborales

def R3(x,y):
  return  y  # Cantidad de horas laborales

# Agregar función objetivo al problema
Q += F(x,y), "Ganancia por venta de ambos modelos"

# Agregar restricciones al problema
Q += R1(x,y) <= 54000, "[R1] Capacidad de la planta"
Q += R2(x,y) <= 20000, "[R2] Disponibilidad de puertas"
Q += R3(x,y) <= 3500, "[R3] Cantidad maxima del modelo"

# Solucionar el problema
Q.solve()
print('\033[1m' + Q.name + '\033[0m')

# Status de la solución encontrada (Unfeasible, Feasible, Optimal)
print("Status:", pulp.LpStatus[Q.status])

# Mostrar la solución óptima
for variables in Q.variables():
    print(f"{variables.name} = {variables.varValue: .0f} [autos/semana]")

# Mostrar el valor de la función objetivo en el óptimo
print(f"Ganancia por venta de autos = {pulp.value(Q.objective)/1E6: 0.2f} [MUSD/semana]")
```
