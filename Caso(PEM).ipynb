### **Ejemplo 4:** Programación Entera Mixta (PEM)

```
# Instalación del peaquete de programación lineal PuLP
!pip install PuLP

# Importar paquete de programación lineal
import pulp
```

```
# Definición del problema (minimización)
Q = pulp.LpProblem("Distribución de bienes a través de una red", pulp.LpMinimize)

# Definir variables de decisión como enteras
xF1_F2 = pulp.LpVariable("xF1_F2", lowBound=0, cat='Integer')
xF1_CD = pulp.LpVariable("xF1_CD", lowBound=0, cat='Integer')
xF1_A1 = pulp.LpVariable("xF1_A1", lowBound=0, cat='Integer')
xF2_CD = pulp.LpVariable("xF2_CD", lowBound=0, cat='Integer')
xCD_A2 = pulp.LpVariable("xCD_A2", lowBound=0, cat='Integer')
xA1_A2 = pulp.LpVariable("xA1_A2", lowBound=0, cat='Integer')
xA2_A1 = pulp.LpVariable("xA2_A1", lowBound=0, cat='Integer')

# Función objetivo
Q += (
    200*xF1_F2 +
    400*xF1_CD +
    900*xF1_A1 +
    300*xF2_CD +
    100*xCD_A2 +
    300*xA1_A2 +
    200*xA2_A1
), "Costo total de envío"

# Restricciones
Q += xF1_F2 + xF1_CD + xF1_A1 == 50, "[R1] Fabrica_1"
Q += -xF1_F2 + xF2_CD == 40, "[R2] Fabrica_2"
Q += -xF1_CD - xF2_CD + xCD_A2 == 0, "[R3] Centro de distribución"
Q += -xF1_A1 + xA1_A2 - xA2_A1 == -30, "[R4] Almacén 1"
Q += -xCD_A2 - xA1_A2 + xA2_A1 == -60, "[R5] Almacén 2"
Q += xF1_F2 <= 10, "[R6] Límite F1->F2"
Q += xCD_A2 <= 80, "[R7] Límite CD->A2"

# Resolver el problema
Q.solve()

# Mostrar resultados
print('\033[1m' + Q.name + '\033[0m')
print("Status:", pulp.LpStatus[Q.status])

# Mostrar valores de variables
for variable in Q.variables():
    print(f"{variable.name} = {variable.varValue:.0f} Productos")

# Mostrar valor de la función objetivo
print(f"Costo_total= {pulp.value(Q.objective)/1E3:.3f} [MUSD]")
```
